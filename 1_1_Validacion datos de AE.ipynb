```python
import numpy as np
df_indicadores_validacion = {'epsilon+':[],'igd+':[],'r2':[]}

for indicador, folder in tqm(list(nombre_folders.items())):

    path_base = f'../../{folder}'

    w_dir = [f'w_0{i}' for i in range(0,11)]
    wi = w_dir[0] 

    dicc_a_df = []

    directorio_w= os.path.join(path_base, wi)
    for file in os.listdir(directorio_w):
        if not file.endswith('pof'):
            continue

        n_objt_file = re.findall(pattern=r'0(\d)D_',string=file)[0]
        directorio_archivo = os.path.join(directorio_w,file)
        df_sol = get_sol_path(path = directorio_archivo, n_objetivos= int(n_objt_file))
        nulos = df_sol.isna().sum().sum()
        infinitos = df_sol.astype(float).replace([np.inf, -np.inf], np.nan).isna().sum().sum()
        dicc_a_df.append({'objetivos_leidos': len(df_sol.columns),'entradas_por_archivo':len(df_sol),'objetivos_reales': n_objt_file[0],'nulos': nulos, 'infinitos': infinitos})

    df_validacion = pd.DataFrame(dicc_a_df)
    df_indicadores_validacion[indicador] =df_validacion
    
        
```

```python
for indicador, df_validacion in df_indicadores_validacion.items():
    print('Indicador:', indicador)
    print('Longitud de las tablas:', len(df_validacion))
    print('Valores únicos de entradas por achivo:', df_validacion['entradas_por_archivo'].unique())
    print('Valores nulos de entradas por achivo:', df_validacion['nulos'].unique())
    print('Valores infinitos de entradas por achivo:', df_validacion['infinitos'].unique())
    print('Columnas iguales  en columnas', (df_validacion['objetivos_leidos']==df_validacion['objetivos_reales'].astype(int)).mean())
```
